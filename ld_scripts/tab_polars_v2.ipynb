```python
# new function
def icd_to_stage(icd):
    """
        'N18.1%': 1,
        'N18.2%': 2, 
        'N18.3%': 3, 
        'N18.4%': 4, 
        'N18.5%': 5, 
        'N18.6%': 'ESRD', 
        'N18.9%': 'CKD'
    """
    if icd is None:
        return (None, 0)
    if icd in 'N18.1%':
        return ("1", 1)
    if icd in 'N18.2%':
        return ("2", 2)
    if icd in 'N18.3%':
        return ("3", 3)
    if icd in 'N18.4%':
        return ("4", 4)
    return ("5", 5)
```
